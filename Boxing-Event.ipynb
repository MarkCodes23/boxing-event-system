```python
events = []
boxers = []

def add_event():
  print("\nAdding a new event...")
  title = input("Enter the event title: ")
  date = input("Enter the event date: ")
  time = input("Enter the event time: ")
  location = input("Enter the event location: ")

  event = {
      "title": title,
      "date": date,
      "time": time,
      "location": location,
      "boxers": []
  }
  events.append(event)
  print("Event added successfully!")

def view_events():
  print("\nViewing all events...")
  if not events:
    print("No events scheduled.")
    return
  for idx, event in enumerate(events):
    print(f"[{idx}] Title: {event['title']}")
    print(f"    Date: {event['date']}")
    print(f"    Location: {event['location']}")
    if event['boxers']:
      boxer_names = ", ".join([boxer['name'] for boxer in event['boxers']])
      print(f"    Boxers: {boxer_names}")
    else:
      print("    Boxers: None")
    print("-" * 20)
  print()

def add_boxer():
  print("\nAdding a new boxer...")
  name = input("Enter the boxer's name: ")
  age = input("Enter the boxer's age: ")
  weightclass = input("Enter the boxer's weight class: ")
  record = input("Enter the boxer's record: ")

  boxer = {
      "name": name,
      "age": age,
      "weightclass": weightclass,
      "record": record
  }
  boxers.append(boxer)
  print("Boxer added successfully!")

def view_boxers():
  print("\nViewing all boxers...")
  if not boxers:
    print("No boxers registered.")
    return
  for idx, boxer in enumerate(boxers):
    print(f"[{idx}] Name: {boxer['name']}")
    print(f"    Age: {boxer['age']}")
    print(f"    Weightclass: {boxer['weightclass']}")
    print(f"    Record: {boxer['record']}")
    print("-" * 20)
  print()

def link_boxer_to_event():
  print("\nLinking a boxer to an event...")
  if not events:
    print("No events available. Please create an event first.")
    return
  if not boxers:
    print("No boxers available. Please add a boxer first.")
    return

  print("Available Events:")
  view_events()
  try:
    event_idx = int(input("Enter the index of the event: "))
    if event_idx < 0 or event_idx >= len(events):
      print("Invalid event index.")
      return
  except ValueError:
    print("Please enter a valid number.")
    return

  print("Available Boxers:")
  view_boxers()
  try:
    boxer_idx = int(input("Enter the index of the boxer: "))
    if boxer_idx < 0 or boxer_idx >= len(boxers):
      print("Invalid boxer index.")
      return
  except ValueError:
    print("Please enter a valid number.")
    return

  selected_event = events[event_idx]
  selected_boxer = boxers[boxer_idx]

  if selected_boxer in selected_event['boxers']:
    print(f"{selected_boxer['name']} is already linked to this event.")
  else:
    selected_event['boxers'].append(selected_boxer)
    print(f"Successfully linked {selected_boxer['name']} to {selected_event['title']}!")

def display_menu():
  while True:
    print("\n--- Boxing Event Manager ---")
    print("1. Add Event")
    print("2. View Events")
    print("3. Add Boxer")
    print("4. View Boxers")
    print("5. Link Boxer to Event")
    print("6. Exit")

    input_choice = input("Enter your choice (1-6): ")
    if input_choice == "1":
      add_event()
    elif input_choice == "2":
      view_events()
    elif input_choice == "3":
      add_boxer()
    elif input_choice == "4":
      view_boxers()
    elif input_choice == "5":
      link_boxer_to_event()
    elif input_choice == "6":
      print("Exiting program.")
      break
    else:
      print("Invalid choice. Please try again.")

if __name__ == "__main__":
  display_menu()
```
